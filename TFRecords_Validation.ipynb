```python
# =========================
# Validate ALL TFRecords (Yeh-style strict, ZERO tolerance)
# - Draw bad clusters as red circles
# - Show % of good vs bad clusters in title
# - Render maps inline as output
# =========================
!pip -q install geopandas shapely pyproj requests

import os, re, json, io, zipfile, requests
from pathlib import Path
from collections import defaultdict

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import geopandas as gpd
import tensorflow as tf
from tqdm.auto import tqdm

# ---------- CONFIG ----------
TFRECORD_DIR = "/content/drive/My Drive/dhs_tfrecords_raw"
CSV_PATH     = "/content/drive/My Drive/Master_Thesis/Surveys/clusters_yeh_spec.csv"
OUT_ROOT     = "/content/drive/My Drive/Master_Thesis/_tfrecord_validation_after"
PLOTS_DIR    = os.path.join(OUT_ROOT, "plots")

REQUIRED_KEYS = [
    "BLUE","GREEN","RED","NIR","SWIR1","SWIR2","TEMP1","LON","LAT","NIGHTLIGHTS",
    "cluster_index","lon","wealthpooled","year"
]
NON_NEGATIVE_BANDS = ["RED","BLUE","GREEN","NIGHTLIGHTS","NIR","SWIR1","SWIR2","TEMP1"]

# ---------- LOAD CLUSTERS CSV ----------
clusters = pd.read_csv(CSV_PATH, float_precision="high", low_memory=False)
clusters = clusters.rename(columns={"cluster": "cluster_index"})
need_cols = {"country", "year", "cluster_index", "lat", "lon"}
missing_cols = need_cols - set(clusters.columns)
if missing_cols:
    raise ValueError(f"{CSV_PATH} missing columns: {sorted(missing_cols)}")

if "wealthpooled" not in clusters.columns:
    clusters["wealthpooled"] = clusters.get("mean_pca_wealth", 0.0)

clusters["cluster_index"] = pd.to_numeric(clusters["cluster_index"], errors="coerce").astype("Int64")
clusters["year"]          = pd.to_numeric(clusters["year"], errors="coerce").astype("Int64")
clusters["lat"]           = pd.to_numeric(clusters["lat"], errors="coerce")
clusters["lon"]           = pd.to_numeric(clusters["lon"], errors="coerce")

# ---------- DISCOVER TFRECORD FILES ----------
pat = re.compile(r"^([A-Z]{2})_(\d{4})_.*\.tfrecord\.gz$")
files = []
for f in os.listdir(TFRECORD_DIR):
    m = pat.match(f)
    if m:
        cc, yr = m.group(1), int(m.group(2))
        files.append((cc, yr, os.path.join(TFRECORD_DIR, f)))
files = sorted(files, key=lambda x: (x[0], x[1], x[2]))
print(f"Found {len(files)} TFRecord files.")

Path(OUT_ROOT).mkdir(parents=True, exist_ok=True)
Path(PLOTS_DIR).mkdir(parents=True, exist_ok=True)

# ---------- HIGH-DETAIL BORDERS ----------
NE_URL = "https://github.com/nvkelso/natural-earth-vector/archive/refs/heads/master.zip"
NE_DIR = "/content/ne_data_10m"
NE_ZIP = "/content/ne_data_10m.zip"

def ensure_natural_earth():
    if os.path.exists(NE_DIR): return
    print("Downloading Natural Earth (1:10m)…")
    r = requests.get(NE_URL, timeout=120); r.raise_for_status()
    with open(NE_ZIP, "wb") as f: f.write(r.content)
    with zipfile.ZipFile(NE_ZIP, "r") as z: z.extractall(NE_DIR)

def load_admin0_10m():
    ensure_natural_earth()
    for root, _, fs in os.walk(NE_DIR):
        if "ne_10m_admin_0_countries.shp" in fs:
            return gpd.read_file(os.path.join(root, "ne_10m_admin_0_countries.shp"))
    raise FileNotFoundError("Natural Earth admin_0 countries (10m) shapefile not found.")

world10 = load_admin0_10m()
ISO_A2_COL = next((c for c in world10.columns if str(c).upper().startswith("ISO_A2")), None)

ISO_OVERRIDES = {"LB": "LR", "MD": "MG"}

def get_country_geom(cc):
    cc2 = ISO_OVERRIDES.get(cc, cc)
    if ISO_A2_COL and cc2 in world10[ISO_A2_COL].astype(str).values:
        g = world10[world10[ISO_A2_COL] == cc2]
        return g if not g.empty else world10.iloc[[]]
    return world10.iloc[[]]

# ---------- HELPERS ----------
def contains_neg(fmap, band):
    arr = np.asarray(fmap[band].float_list.value, dtype=np.float32)
    return np.any(arr < 0)

def plot_good_bad(cc, yr, df_csv, bad_clusters, out_png, wealth_col="wealthpooled"):
    df = df_csv[["cluster_index", wealth_col, "lat", "lon"]].dropna().copy()
    df["cluster_index"] = df["cluster_index"].astype(int)
    bad_set = set(int(x) for x in bad_clusters)
    good = df.loc[~df["cluster_index"].isin(bad_set)]
    bad  = df.loc[df["cluster_index"].isin(bad_set)]

    total = len(df)
    n_good, n_bad = len(good), len(bad)
    p_good = (n_good / total * 100) if total > 0 else 0
    p_bad  = (n_bad / total * 100) if total > 0 else 0

    fig, ax = plt.subplots(figsize=(7,7))
    geom = get_country_geom(cc)
    if not geom.empty:
        geom.boundary.plot(ax=ax, linewidth=0.8, color="black")

    if len(good):
        vmin, vmax = np.nanpercentile(good[wealth_col], [1, 99])
        sc = ax.scatter(good["lon"], good["lat"], c=good[wealth_col],
                        s=8, vmin=vmin, vmax=vmax, cmap="viridis", label=f"Good ({p_good:.1f}%)")
        cbar = plt.colorbar(sc, ax=ax); cbar.set_label(wealth_col)
    if len(bad):
        ax.scatter(bad["lon"], bad["lat"], edgecolor="red", facecolor="none",
                   s=40, linewidths=1.4, label=f"Bad ({p_bad:.1f}%)")

    ax.set_title(f"{cc}-{yr}: Good vs Bad (ZERO tol)  |  Good: {p_good:.1f}%  Bad: {p_bad:.1f}%")
    ax.set_xlabel("Longitude"); ax.set_ylabel("Latitude")
    ax.legend(); ax.set_aspect("equal","box")
    plt.tight_layout()
    Path(os.path.dirname(out_png)).mkdir(parents=True, exist_ok=True)
    plt.savefig(out_png, dpi=220, bbox_inches="tight")

    # Display inline in notebook
    plt.show()
    plt.close(fig)

# ---------- MAIN ----------
by_survey = defaultdict(list)
for cc, yr, path in files:
    by_survey[(cc, yr)].append(path)

overall = []
for (cc, yr), paths in tqdm(sorted(by_survey.items()), desc="Validating surveys"):
    sub = clusters.query("country == @cc and year == @yr").copy()
    if sub.empty:
        print(f"Skip {cc}-{yr}: no matching rows in CSV."); continue

    csv_clusters = set(sub["cluster_index"].dropna().astype(int))
    lon_lookup = dict(zip(sub["cluster_index"].astype(int), sub["lon"].astype(np.float32)))
    wealth_col = "wealthpooled" if "wealthpooled" in sub.columns else "mean_pca_wealth"

    rejects, missing = [], set(csv_clusters)
    good_cnt, seen_cnt = 0, 0

    for path in paths:
        try:
            ds = tf.data.TFRecordDataset(path, compression_type="GZIP")
        except Exception as e:
            rejects.append({"COUNTRY": cc, "YEAR": yr, "CLUSTER": None,
                            "reason": f"open_failed:{e}", "file": os.path.basename(path)})
            continue

        for raw in ds:
            seen_cnt += 1
            try:
                ex = tf.train.Example.FromString(raw.numpy())
                fmap = ex.features.feature
            except Exception as e:
                rejects.append({"COUNTRY": cc, "YEAR": yr, "CLUSTER": None,
                                "reason": f"parse_failed:{e}", "file": os.path.basename(path)})
                continue

            # 1) Required keys
            if any(k not in fmap for k in REQUIRED_KEYS):
                missing_k = [k for k in REQUIRED_KEYS if k not in fmap]
                rejects.append({"COUNTRY": cc, "YEAR": yr, "CLUSTER": None,
                                "reason": f"missing_required_keys:{missing_k}",
                                "file": os.path.basename(path)})
                continue

            # 2) Extract props
            try:
                clus = int(fmap["cluster_index"].float_list.value[0])
                year_val = int(fmap["year"].float_list.value[0])
                lon_prop = np.float32(fmap["lon"].float_list.value[0])
            except Exception as e:
                rejects.append({"COUNTRY": cc, "YEAR": yr, "CLUSTER": None,
                                "reason": f"invalid_fields:{e}", "file": os.path.basename(path)})
                continue

            if year_val != yr or clus not in csv_clusters:
                rejects.append({"COUNTRY": cc, "YEAR": yr, "CLUSTER": clus,
                                "reason": "bad_year_or_cluster", "file": os.path.basename(path)})
                continue

            exp_lon = np.float32(lon_lookup[clus])
            if lon_prop != exp_lon:
                rejects.append({"COUNTRY": cc, "YEAR": yr, "CLUSTER": clus,
                                "reason": f"lon_mismatch:{lon_prop} vs {exp_lon}",
                                "file": os.path.basename(path)})
                continue

            if any(contains_neg(fmap, b) for b in NON_NEGATIVE_BANDS):
                rejects.append({"COUNTRY": cc, "YEAR": yr, "CLUSTER": clus,
                                "reason": "negative_values", "file": os.path.basename(path)})
                continue

            if clus in missing: missing.remove(clus)
            good_cnt += 1

    # Save per-survey outputs
    rej_df = pd.DataFrame(rejects)
    rej_csv = os.path.join(OUT_ROOT, f"{cc}_{yr}_rejects.csv")
    rej_df.to_csv(rej_csv, index=False)

    miss_df = pd.DataFrame(sorted(missing), columns=["cluster_index"])
    miss_df.insert(0, "country", cc); miss_df.insert(1, "year", yr)
    miss_csv = os.path.join(OUT_ROOT, f"{cc}_{yr}_missing.csv")
    miss_df.to_csv(miss_csv, index=False)

    plot_png = os.path.join(PLOTS_DIR, f"{cc}_{yr}.png")
    plot_good_bad(cc, yr, sub, bad_clusters=missing, out_png=plot_png, wealth_col=wealth_col)

    overall.append({
        "country": cc, "year": yr,
        "files": len(paths),
        "examples_seen": seen_cnt,
        "good": good_cnt,
        "bad": int(len(rej_df)),
        "missing": int(len(missing)),
        "rejects_csv": rej_csv,
        "missing_csv": miss_csv,
        "plot": plot_png
    })

# Save summary JSON
summary_path = os.path.join(OUT_ROOT, "summary_all.json")
with open(summary_path, "w") as f:
    json.dump(overall, f, indent=2)

print(f"\nSaved outputs → {OUT_ROOT}\nPlots → {PLOTS_DIR}\nSummary → {summary_path}")
```
